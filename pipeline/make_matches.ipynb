```python

import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np

img1 = mpimg.imread('../images/table1.jpeg')
img2 = mpimg.imread('../images/table2.jpeg')
img1 = np.rot90(img1, k=-1)
img2 = np.rot90(img2, k=-1)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))
## plot as portrait and not landscape
ax1.imshow(img1)
ax2.imshow(img2)

ax1.set_title('Image 1 — click points')
ax2.set_title('Image 2 — click points')

points1, points2 = [], []
current = [1]  # which image we're clicking

def onclick(event):
    if event.inaxes == ax1:
        points1.append((event.xdata, event.ydata))
        ax1.plot(event.xdata, event.ydata, 'r+', markersize=12)
        ax1.text(event.xdata, event.ydata, 
                 str(len(points1)), color='red', fontsize=10)
    elif event.inaxes == ax2:
        points2.append((event.xdata, event.ydata))
        ax2.plot(event.xdata, event.ydata, 'b+', markersize=12)
        ax2.text(event.xdata, event.ydata,
                 str(len(points2)), color='blue', fontsize=10)
    fig.canvas.draw()
    print(f'Points1: {len(points1)}, Points2: {len(points2)}')

def onclose(event):
    n = min(len(points1), len(points2))
    correspondences = {
        'points1': points1[:n],
        'points2': points2[:n]
    }
    # with open('correspondences.csv', 'w') as f:
    #     for (x1, y1), (x2, y2) in zip(points1[:n], points2[:n]):
    #         f.write(f'{x1},{y1},{x2},{y2}\n')
    # print(f'Saved {n} correspondences to correspondences.csv')

fig.canvas.mpl_connect('button_press_event', onclick)
fig.canvas.mpl_connect('close_event', onclose)
plt.tight_layout()
plt.show()
```

```python
n = min(len(points1), len(points2))
with open('correspondences.csv', 'w') as f:
    for (x1, y1), (x2, y2) in zip(points1[:n], points2[:n]):
        f.write(f'{x1},{y1},{x2},{y2}\n')
print(f'Saved {n} correspondences to correspondences.csv')
```
